```python
import pandas as pd
from scipy import stats

df = pd.read_csv("2) Stock Prices Data Set.csv")

print("Missing Values:")
print(df[['open', 'close']].isnull().sum())

df_clean = df[['open', 'close']].dropna()

if df_clean['open'].nunique() == 1 or df_clean['close'].nunique() == 1:
    print("One of the columns has a constant value. Regression is not possible")
else:
  slope, intercept, r_value, p_value, std_err = stats.linregress(df_clean['open'], df_clean['close'])
  print("Linear Regression: open vs close")
  print(f"Regression Equation: y = {slope:.2f} * open + ({intercept:.2f})")
  print(f"Correlation Coefficient (r): {r_value:.3f}")
  print(f"P-value: {p_value:.1e}")
```
